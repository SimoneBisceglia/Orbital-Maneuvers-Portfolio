```python
import math
import numpy as np
import matplotlib.pyplot as plt
import orekit
from orekit.pyhelpers import setup_orekit_curdir

# ==========================================
# 1. OREKIT ENVIRONMENT INITIALIZATION
# ==========================================
orekit.initVM()
setup_orekit_curdir('C:/Users/Simone/Downloads/orekit-data-main')

from org.orekit.propagation.analytical.tle import TLE, TLEPropagator
from org.orekit.frames import FramesFactory
from org.orekit.time import TimeScalesFactory

# ==========================================
# 2. TLE INPUT (ENVISAT 15/06/2024 - CelesTrak)
# ==========================================
line1 = "1 27386U 02009A   24167.89209538  .00000181  00000-0  00000-0 0  9991"
line2 = "2 27386  98.4005 125.3855 0001397  89.4312 270.7067 14.38002035165625"
tle = TLE(line1, line2)

# ==========================================
# 3. SGP4 ANALYTICAL PROPAGATOR CREATION
# ==========================================
sgp4_propagator = TLEPropagator.selectExtrapolator(tle)

# ==========================================
# 4. INITIAL STATE EXTRACTION (TLE Epoch)
# ==========================================
initial_state = sgp4_propagator.getInitialState()
epoch = initial_state.getDate()

# Coordinates converted into the EME2000 Inertial Frame (ECI)
inertial_frame = FramesFactory.getEME2000()
pv_coordinates = initial_state.getPVCoordinates(inertial_frame)

# Extraction in meters
pos = pv_coordinates.getPosition()
vel = pv_coordinates.getVelocity()

x_m, y_m, z_m = pos.getX(), pos.getY(), pos.getZ()
vx_ms, vy_ms, vz_ms = vel.getX(), vel.getY(), vel.getZ()

# --- DATE FORMATTING ---
epoch_str = str(epoch)[:19].replace('T', '    at time    ')

# ==========================================
# 5. CONSOLE OUTPUT (Clean & Professional)
# ==========================================
print("\n" + "="*65)
print("    SGP4: ECI STATE VECTOR EXTRACTION")
print("="*65 + "\n")

print(f"    Date & Time (UTC) :  {epoch_str}\n")
print(f"    Reference Frame   :  EME2000 (Inertial)\n")

print("-" * 65 + "\n")

print("    POSITION [m] :")
print(f"        X  =  {x_m:18.4f}")
print(f"        Y  =  {y_m:18.4f}")
print(f"        Z  =  {z_m:18.4f}\n")

print("    VELOCITY [m/s] :")
print(f"        Vx =  {vx_ms:18.4f}")
print(f"        Vy =  {vy_ms:18.4f}")
print(f"        Vz =  {vz_ms:18.4f}\n")

print("="*65 + "\n")
```
